LOOPS

```python
# While  loop
i = 1 # intilalize
while(i <= 10): #condition
    print(i)
    i += 1 #increment
```

```python
# EX - 2: break statement
# break statement helps / control to come out of the loop

i = 1
while (i <= 10):
    print(i)
    if i == 5:
        break
    i += 1
```

```python
# EX- 3: Continue statement
# Continue will take the control to the start of the loop.
# if continue will execute then further statements will not continue

i = 1
while i < 10:
    i += 1
    print(i)
    if i < 5:
        continue
    
```

```python
i = 1
while i < 10:
    if i < 5:
        continue
    i += 1
    print(i)
```

```python
i = 1
while i < 10:
    i += 1
    if i < 5:
        continue
    
```

```python
# Ex - 5: pass statement
# pass statemnt is a placeholder for furture code

i = 0
while i <= 10:
    if i == 5:
        pass
    else:
        print(i)
    i += 1
```

for loop


```python
animals = ["puli","pilli","kukka","pandhi","kothi"]
for i in animals:
    print(i)
```

```python
food_name = "CHICKEN_DUM_BIRYANI"
for i in food_name:
    print(i)
```

```python
# for loop using with range()
#range() is used to generate list of elements..
#Syntax of range : range(start, end, step)
#start = 0
#end = end - 1
#step = 1
for i in range(4,10,2): 
    print(i)
    
```

```python
# EX : break statement
#Break statement will bring the control out of the loop

for num in range(1,6):
    if num == 3:
        break #it cancome out of loop
    print(num)
```

```python
# EX : continue statement -take control to the start of the loop
for num in range(1,6):
    if num == 3:
        continue
    print(num)
```

```python
# EX nested loop
# for every value of outer loop, inner loop will completely execute..
for i in range(3, 6): #[1,2,3 n = times
    for j in range(2, 7): #[1,2] m = 5 times
        print(i,j) #time complexity n*m
        
```

```python
# EX - else statement

for i in range(5):
    if i == 2:
        continue
    elif i == 40:
        break
    else:
        print(i)
    pass
else:
    print("loop completed successfully")
    
```

```python
# Ex - pattern promblems
for i in range(1,6): 
    for j in range(1,6): 
        print("*", end = " ")
    print()
```

```python
for i in range(1,6): 
    for j in range(i): 
        print("*", end = " ")
    print()
```

```python
for i in range(6,0,-1): 
    for j in range(i): 
        print("*", end = " ")
    print()
```

```python
n = 6
for i in range(n,0,-1):
    print(" " * i + "* " * (n-i))
          # Spaces     #no of stars

```
