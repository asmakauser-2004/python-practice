## operations on strings

```python
# 1. Concatenation (+)
# 2. Repetition(*)
# 3. Membership(in ,not in)
# 4. Indexing & Slicing
```

### 1.Concatentation(+)

```python
str1 = " Babai "
str2 = " likes "
str3 = " pinni "
str4 = str1 + str2 + str3
print(str4)
```

### 2. Repetition(*)

```python
str5 = " bla "
print(10 * str5)
```

### 3.Membership(in,not in)

```python
str6 = "Hello World"
print("Hell" in str6)
print("Helll" in str6)
print("Helo" not in str6)
```

### 4. Indexing & Slicing

```python
Heroine = "Samantha"
print(Heroine[4])
print(Heroine[-4])
print(Heroine[0:5])
print(Heroine[0:5:2])
print(Heroine[::])
print(Heroine[6:3:-1])
print(Heroine[-1:-6: -1])
print(Heroine[-7::])
print(Heroine[-1:-5:1])
print(Heroine[-1:-6:-2])
print(Heroine[0:2:1] + Heroine[5::1])
print(Heroine[::-1])
```

```python
## reverse the string 
print(Heroine[-1:-9:-1])
```

```python
Heroine = "Samantha"
rev_result= ""
for char in Heroine:
    rev_result = char + rev_result
print(rev_result)    
```
