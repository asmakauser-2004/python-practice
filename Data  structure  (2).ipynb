## updating list item

```python
list_politician= ["Vijay","CBN","Pavan","Jailatha","Balayya","Anushka"]
list_politician[1] = "Dance with"
print(list_politician)
```

```python
# update multiple elements : update more elements with less
list_politician[3:5]=["Dance with"]
print(list_politician)
```

```python
# update less elements with more
list_politician[1:4] = ["Asma","Abdul","Kauser","Kalam"]
print(list_politician)
```

```python
# update less elements with more
list_politician[-5:-1] =  ["Dance with"]
print(list_politician)
```

##  delete item from list (remove(),pop(),clear(),del)

```python
food_list =["papu","Biryani","fried rice ","sambar","payasum","curd rice"]
```

```python
# remove() is used for delete the item with given name
food_list =["papu","Biryani","fried rice","sambar","payasum","curd rice"]
food_list.remove("fried rice")
print(food_list)
food_list.pop()
print(food_list)
```

```python
food_list =["papu","Biryani","fried rice","sambar","payasum","curd rice"]
# delete multiple elements together
del food_list[1:4]
print(food_list)
```

```python
# clear() is used to delete all the items from the list
food_list.clear()
print(food_list)
```

```python
# delete the list itself 
del food_list
print(food_list)
```

# list comprehension

```python
# creating the list in a single line using a loop

l1 = [i for i in range(1,11)]
print(l1)
```

```python
l1 = [i**2  for i in range(1,11)]
print(l1)
```

```python
l1 = [i for i in range(1,11) if i%2==0]
print(l1)
```

```python
l1 = [i for i in range(1,11) if i%2==0 if i%3==0] 
print(l1)
```

```python
for i in range(1,11): 
    if i%2==0: 
         if i%3==0:
           print(i,end = " ")
```

```python
l4 = [i**2 if i%2==0 else i+2 for i in range (1,11)]
print(l4)
```

```python
for i in range(1,11):
    if i%2==0:
        print(i**2)
    else:
        print(i+2)
```

```python
l4 = [i**2 if i%2==0 else i+2 for i in range (1,11) if i>5]
print(l4)
```

```python
l4 = [i**2 if i%2==0 else i+2 for i in range (1,11) if i>5 and i<9]
print(l4)
```

```python
for i in range (1,11):
    if i>5 and i<9 :
       if i%2 == 0:
          print(i**2)
       else:
          print(i+2)
```
