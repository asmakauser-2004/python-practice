# match case

```python
# match case is used to replace multiple if else condition
# match case is more powerful than switch case statement 
# match case works with variable and data structures.
```

```python
# Example-1
num = int(input("Enter a number"))
match num:
    case 10:
        print("you entered 10")
    case 20:
        print("you entered 20")
    case 30:
        print("you entered 30")
    case _:  
        print("Something else")
```

```python
Biryani = input("Famous  Hyderbadi Biryani")
match Biryani:
    case "Paradise":
        print("Spicy & Tasty")
    case "Shahdad":
        print("Double masala")
    case "road side":
        print("Budget friendly")
    case _: 
        print("PG Biryani")
```

```python
# Example-2: passing data structure to match case
data = [1,2,3,4]
match data:
    case [4,3,2,1]:
        print("Hello")
    case [1,2,3,4]:
        print("World")
```

```python
# Example-3 : if condition with match case
val = int(input("Enter a number"))
match val:
    case x if x > 0:
        print("It is a postive number ")
    case y if y < 0:
        print("It is a negative number")
    case _:
        print("It is a Zero")
```
